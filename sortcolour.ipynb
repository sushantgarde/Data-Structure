```python
nums = [0,0,2,2,1,1,0]
```

# initializing the 3 list 0,1,2 respectively

```python
low = []
mediun = []
high = []
```

# use for loop for traverse the list 
# then use if stmt for 0 insert into list low, 1 for medium and 2 for high 

```python
for i in range(len(nums)):
    if nums[i] == 0:
        low.append(0)
    elif nums[i] == 1:
        mediun.append(1)
    elif nums[i] == 2:
        high.append(2)
```

# Just adding the 3 list 

```python
nums = low+mediun+high
```
